--call--魔术方法让以一个类的实例对象可以像函数一样被直接调用


```python
class Adder:

  def __init__(self, n):
    self.n = n

  def __call__(self, x):
    return x + self.n


add_5 = Adder(5)
print(add_5(10))  # 输出 15，对象像函数一样被调用
```
